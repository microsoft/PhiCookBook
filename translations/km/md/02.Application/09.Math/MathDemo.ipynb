## នាំចូលបណ្ណាល័យដែលត្រូវការ
នាំចូល PyTorch និងបណ្ណាល័យ transformers ដែលចាំបាច់សម្រាប់ទាញយក និងប្រើម៉ូដែល Phi-4។


```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
```

## កំណត់គ្រាប់លេខចៃដន្យ
កំណត់គ្រាប់លេខចៃដន្យ ដើម្បីធានាលទ្ធផលអាចធ្វើឡើងវិញបានលើការរត់ផ្សេងៗ


```python
torch.random.manual_seed(0)
```

## ផ្ទុកម៉ូដែល Phi-4-mini-flash-reasoning និង Tokenizer
ផ្ទុក Microsoft Phi-4-mini-flash-reasoningmodel និង tokenizer ដែលពាក់ព័ន្ធពី Hugging Face។ ម៉ូដែលនេះនឹងត្រូវដាក់​នៅលើ CUDA ដើម្បីធ្វើការព្យាករណ៍ដោយមានល្បឿនលឿន។


```python
model_id = "microsoft/Phi-4-mini-flash-reasoning"
model = AutoModelForCausalLM.from_pretrained(
   model_id,
   device_map="cuda",
   torch_dtype="auto",
   trust_remote_code=True,
)
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

## រៀបចំសារ​បញ្ចូល
បង្កើតសារ​សន្ទនា​មួយ​ដែល​មាន​បញ្ហាគណិតវិទ្យា​សមីការថ្នាក់ទីពីរ ហើយទុកទ្រង់ទ្រាយវា​ដោយប្រើ​ទំរង់សន្ទនា​សម្រាប់ម៉ូឌែល។


```python
messages = [{
   "role": "user",
   "content": "How to solve 3*x^2+4*x+5=1?"
}]   
inputs = tokenizer.apply_chat_template(
   messages,
   add_generation_prompt=True,
   return_dict=True,
   return_tensors="pt",
)
```

## បង្កើតចម្លើយ
បង្កើតចម្លើយពីម៉ូដែលដោយប្រើប៉ារ៉ាម៉ែត្រដែលបានបញ្ជាក់ ដូចជា temperature និង top_p ដើម្បីគ្រប់គ្រងភាពចៃដន្យក្នុងលទ្ធផល។


```python
outputs = model.generate(
   **inputs.to(model.device),
   max_new_tokens=32768,
   temperature=0.6,
   top_p=0.95,
   do_sample=True,
)
```

## បកស្រាយលទ្ធផលទៅជាអត្ថបទ

បម្លែងលំដាប់តូកិនដែលបានបង្កើតឡើងវិញទៅជាអត្ថបទដែលមនុស្សអាចអានបាន, ដោយមិនរួមបញ្ចូលតូកិនបញ្ចូលដើម ដើម្បីបង្ហាញតែចម្លើយរបស់ម៉ូឌែល។


```python
outputs = tokenizer.batch_decode(outputs[:, inputs["input_ids"].shape[-1]:])
```
